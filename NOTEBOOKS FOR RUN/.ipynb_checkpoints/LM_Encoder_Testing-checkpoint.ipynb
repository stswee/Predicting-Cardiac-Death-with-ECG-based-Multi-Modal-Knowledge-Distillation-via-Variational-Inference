```python
# biobert.py

# Function to extract text starting from 'Assessment and Plan'
def extract_assessment_and_plan(text):
    # Find the position of 'Assessment and Plan' and extract the text from that point onward
    start_index = text.lower().find("assessment and plan")
    if start_index != -1:
        return text[start_index:]
    else:
        return ""

# Function to get text embeddings
def get_embedding(text):
    inputs = tokenizer(text, return_tensors="pt", truncation=True, padding=True)
    with torch.no_grad():
        outputs = model_lm(**inputs)
    # Get the [CLS] token embedding (first token)
    embedding = outputs.last_hidden_state[:, 0, :].squeeze().numpy()  # Remove batch dimension
    return embedding

# Load BioBERT tokenizer and model
model_name = "dmis-lab/biobert-v1.1"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model_lm = AutoModel.from_pretrained(model_name)

# Get clinical note
df = pd.read_csv("../Data/subject-info-cleaned-with-prompts-and-notes-combined-deaths_Llama8B.csv")
df['Assessment'] = df['Reports'].apply(extract_assessment_and_plan)

# Use tqdm to monitor the progress of applying the function to the DataFrame
tqdm.pandas(desc="Processing clinical reports")

# Apply the function with tqdm progress bar
df['embedding'] = df['Assessment'].progress_apply(get_embedding)

# Now, create a new DataFrame with patient ID and corresponding embeddings
embedding_df = df[['Patient ID', 'embedding']]
```

```python
# datapreprocessing.py
from sklearn.model_selection import train_test_split
import numpy as np
import pandas as pd
import ast
import torch

# Get data
df = pd.read_csv("../Data/subject-info-cleaned.csv")

# Combine embeddings and patient outcome
df2 = pd.concat([embedding_df['embedding'], df['Cause of death']], axis = 1)

# Perform train test split
train_df, test_df = train_test_split(df2, test_size = 0.5, random_state = 42, stratify = df2['Cause of death'])

# Get test values
X_test = np.vstack(test_df['embedding'])
X_test = torch.tensor(X_test)
y_test = test_df['Cause of death']
y_test = np.where(y_test == 0, 0, 1)
y_test = torch.tensor(y_test)
y_test = y_test.to(torch.float64)

### Perform downsampling ###

# Compute class distribution
class_counts = train_df['Cause of death'].value_counts()
min_class_size = class_counts.min()

# Assign weights
weights = train_df['Cause of death'].map(lambda x: 1/ class_counts[x])
weights /= weights.sum() 

# Perform weighted downsampling
downsampled_df = train_df.sample(n = min_class_size * len(class_counts), weights = weights, random_state = 42)

# Get training values
X_train = np.vstack(downsampled_df['embedding'])
X_train = torch.tensor(X_train)
y_train = downsampled_df['Cause of death']
y_train = np.where(y_train == 0, 0, 1)
y_train = torch.tensor(y_train)
y_train = y_train.to(torch.float64)

# # Get embeddings
# X = np.vstack(embedding_df['embedding'])
# # Convert to tensor
# X = torch.tensor(X)

# # Get cause of death
# y = df['Cause of death']
# # Binarize
# y = np.where(y == 0, 0, 1)
# # Convert to tensor
# y = torch.tensor(y)
# y = y.to(torch.float64)

# # Perform train test split
# X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.5, random_state = 42)
```

```python
# models.py
import torch
import torch.nn as nn
import torch.nn.functional as F

# Logistic Regression Model
class LogisticRegressionModel(nn.Module):
    def __init__(self, input_dim):
        super(LogisticRegressionModel, self).__init__()
        self.fc = nn.Linear(input_dim, 1)

    def forward(self, x):
        return self.fc(x)

# MLP Model
class MLPModel(nn.Module):
    def __init__(self, input_dim, hidden_dim=128, num_layers=4, dropout_prob=0.1):
        super(MLPModel, self).__init__()
        layers = []
        
        # First layer (input layer)
        layers.append(nn.Linear(input_dim, hidden_dim))
        layers.append(nn.ReLU())
        layers.append(nn.Dropout(dropout_prob))  # Apply dropout after activation
        
        # Hidden layers
        for _ in range(num_layers - 1):
            layers.append(nn.Linear(hidden_dim, hidden_dim))
            layers.append(nn.BatchNorm1d(hidden_dim))
            layers.append(nn.ReLU())
            layers.append(nn.Dropout(dropout_prob))  # Apply dropout after activation
        
        # Output layer
        layers.append(nn.Linear(hidden_dim, 1))
        
        self.model = nn.Sequential(*layers)

    def forward(self, x):
        return self.model(x)

# Function to select model
def get_model(model_name, input_dim):
    if model_name == 'logistic_regression':
        return LogisticRegressionModel(input_dim)
    elif model_name == 'mlp':
        return MLPModel(input_dim)
    else:
        raise ValueError(f"Unknown model: {model_name}")
```

```python
# Loss functions
# class FocalLoss(nn.Module):
#     def __init__(self, alpha=0.75, gamma=2.0, reduction="mean"):
#         """
#         Implements Focal Loss for binary classification.
#         Args:
#             alpha (float): Balancing factor for class weights.
#             gamma (float): Focusing parameter for hard examples.
#             reduction (str): 'mean' (default) or 'sum'.
#         """
#         super(FocalLoss, self).__init__()
#         self.alpha = alpha
#         self.gamma = gamma
#         self.reduction = reduction
#         self.bce_loss = nn.BCEWithLogitsLoss(reduction="none")  # Keep raw loss for focal scaling

#     def forward(self, logits, targets):
#         """
#         Args:
#             logits: Raw model outputs (before sigmoid).
#             targets: Ground truth labels (binary, 0 or 1).
#         Returns:
#             Focal Loss value.
#         """
#         bce_loss = self.bce_loss(logits, targets)  # Compute standard BCE loss
#         probs = torch.sigmoid(logits)  # Convert logits to probabilities
#         targets = targets.float()

#         # Compute focal loss scaling factor
#         p_t = targets * probs + (1 - targets) * (1 - probs)
#         focal_factor = (1 - p_t) ** self.gamma

#         # Apply alpha weighting (for class balance)
#         alpha_factor = targets * self.alpha + (1 - targets) * (1 - self.alpha)

#         # Compute final focal loss
#         loss = alpha_factor * focal_factor * bce_loss

#         if self.reduction == "mean":
#             return loss.mean()
#         elif self.reduction == "sum":
#             return loss.sum()
#         else:
#             return loss  # No reduction (per-sample loss)

import torch
import torch.nn as nn

class FocalLoss(nn.Module):
    def __init__(self, alpha=0.25, gamma=2.0, class_weights=None, reduction="mean"):
        """
        Implements Focal Loss for binary classification with class weights.
        Args:
            alpha (float): Balancing factor for class weights (can be used if class_weights is not provided).
            gamma (float): Focusing parameter for hard examples.
            class_weights (tuple or list, optional): Weights for class 0 and class 1 (e.g., [weight_0, weight_1]).
            reduction (str): 'mean' (default) or 'sum'.
        """
        super(FocalLoss, self).__init__()
        self.alpha = alpha
        self.gamma = gamma
        self.reduction = reduction
        self.class_weights = class_weights
        self.bce_loss = nn.BCEWithLogitsLoss(reduction="none")  # Keep raw loss for focal scaling

    def forward(self, logits, targets):
        """
        Args:
            logits: Raw model outputs (before sigmoid).
            targets: Ground truth labels (binary, 0 or 1).
        Returns:
            Focal Loss value.
        """
        bce_loss = self.bce_loss(logits, targets)  # Compute standard BCE loss
        probs = torch.sigmoid(logits)  # Convert logits to probabilities
        targets = targets.float()

        # Compute focal loss scaling factor
        p_t = targets * probs + (1 - targets) * (1 - probs)
        focal_factor = (1 - p_t) ** self.gamma

        # Apply class weights (if provided)
        if self.class_weights is not None:
            weight_0, weight_1 = self.class_weights
            weight_factor = targets * weight_1 + (1 - targets) * weight_0
        else:
            # Default behavior if no class_weights are provided
            weight_factor = targets * self.alpha + (1 - targets) * (1 - self.alpha)

        # Compute final focal loss with class weighting
        loss = weight_factor * focal_factor * bce_loss

        if self.reduction == "mean":
            return loss.mean()
        elif self.reduction == "sum":
            return loss.sum()
        else:
            return loss  # No reduction (per-sample loss)
```

```python
# train.py

import torch
import torch.optim as optim
import torch.nn as nn
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, roc_curve
from sklearn.model_selection import KFold, StratifiedKFold
from sklearn.utils.class_weight import compute_class_weight
import matplotlib.pyplot as plt
from tqdm import tqdm
import numpy as np

def set_random_seed(seed=42):
    # Set random seed for Python's built-in random module
    random.seed(seed)
    
    # Set random seed for NumPy
    np.random.seed(seed)
    
    # Set random seed for PyTorch
    torch.manual_seed(seed)
    
    # If using GPU, set the random seed for CUDA as well
    torch.cuda.manual_seed_all(seed)
    
    # Ensure deterministic behavior in PyTorch
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False  # May affect performance but ensures deterministic results

# Compute class weights based on the training labels
def compute_class_weights(y_train):
    # Convert PyTorch tensor to NumPy array
    y_train_np = y_train.cpu().numpy() if y_train.is_cuda else y_train.numpy()
    
    # Compute the class weights using 'balanced' mode
    class_weights = compute_class_weight(class_weight='balanced', classes=np.unique(y_train_np), y=y_train_np)
    
    # Convert class weights to PyTorch tensor
    return torch.tensor(class_weights, dtype=torch.float32).to(y_train.device)

def train_model(X_train, y_train, X_test, y_test, model, k = 3, epochs=100, lr=0.005):

    # Set random seed for reproducibility
    set_random_seed(42)
    
    # Set device to GPU
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

    # Compute class weights based on y_train
    # class_weights = compute_class_weights(y_train)
    class_weights = [0.5, 0.5]

    # Create list of k models for k-fold cross-validation
    model_list = [model for i in range(k)]

    # Keep track of best model
    best_model = None
    best_val_loss = float('inf')
    val_losses = []

    # k-fold cross-validation
    kfold = StratifiedKFold(n_splits = k, shuffle = True, random_state = 42)

    # Iterate through each fold
    for fold, (train_idx, val_idx) in enumerate(kfold.split(X_train, y_train)):
        X_train_fold, y_train_fold = X_train[train_idx], y_train[train_idx]
        X_val_fold, y_val_fold = X_train[val_idx], y_train[val_idx]

        # Initialize model
        model = model_list[fold].to(device)
        criterion = FocalLoss(class_weights = class_weights).to(device)
        optimizer = optim.AdamW(model.parameters(), lr=lr, weight_decay = 1e-3)
        scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=2, gamma=0.5)

        # Training loop
        for epoch in tqdm(range(epochs), desc=f"Fold {fold+1} Training", leave=False):
            model.train()
            optimizer.zero_grad()
            outputs = model(X_train_fold.to(device))
            loss = criterion(outputs.squeeze(), y_train_fold.to(device))
            loss.backward()
            optimizer.step()
            scheduler.step()

        # Evaluate on validation set
        model.eval()
        val_loss = 0.0
        correct = 0
        total = 0
        all_targets = []
        all_predictions = []
        all_probs = []

        batch_size = 32
        with torch.no_grad():
            for i in range(0, len(X_val_fold), batch_size):
                X_batch = X_val_fold[i:i+batch_size].to(device)
                y_batch = y_val_fold[i:i+batch_size].to(device)
                
                outputs = model(X_batch)
                loss = criterion(outputs.squeeze(), y_batch)
                val_loss += loss.item()
                
                probs = torch.sigmoid(outputs).squeeze()
                predicted = torch.round(probs)  # Assuming binary classification
                
                all_targets.extend(y_batch.cpu().numpy())
                all_predictions.extend(predicted.cpu().numpy())
                all_probs.extend(probs.cpu().numpy())
                
                correct += (predicted == y_batch).sum().item()
                total += y_batch.size(0)

        # Validation loss
        avg_val_loss = val_loss / (len(X_val_fold) / batch_size)

        # Update best model
        if avg_val_loss < best_val_loss:
            best_val_loss = avg_val_loss
            best_model_weights = model.state_dict() # Save model weights

        # Keep record of validation losses
        val_losses.append(avg_val_loss)

        # Metrics
        accuracy = correct / total
        precision = precision_score(all_targets, all_predictions)
        recall = recall_score(all_targets, all_predictions)
        f1 = f1_score(all_targets, all_predictions)
        auc = roc_auc_score(all_targets, all_probs)
        
        print(f"Fold {fold+1}: Validation Loss = {avg_val_loss:.4f}, Accuracy = {accuracy:.4f}, Precision = {precision:.4f}, Recall = {recall:.4f}, F1 = {f1:.4f}, AUC = {auc:.4f}")
    
    print("Training complete.")

    # Evaluate on test set
    model.eval()
    test_loss = 0.0
    correct = 0
    total = 0
    all_targets = []
    all_predictions = []
    all_probs = []

    with torch.no_grad():
        X_test = X_test.to(device)
        y_test = y_test.to(device)
    
        outputs = model(X_test)
        loss = criterion(outputs.squeeze(), y_test)
        test_loss += loss.item()

        probs = torch.sigmoid(outputs).squeeze()
        predicted = torch.round(probs)

        all_targets.extend(y_test.cpu().numpy())
        all_predictions.extend(predicted.cpu().numpy())
        all_probs.extend(probs.cpu().numpy())

        correct += (predicted == y_test).sum().item()
        total += y_test.size(0)

    avg_test_loss = test_loss / len(X_test)
    accuracy = correct / total
    precision = precision_score(all_targets, all_predictions)
    recall = recall_score(all_targets, all_predictions)
    f1 = f1_score(all_targets, all_predictions)
    auc = roc_auc_score(all_targets, all_probs)

    print(f"Test Set Results: Loss = {avg_test_loss:.4f}, Accuracy = {accuracy:.4f}, Precision = {precision:.4f}, Recall = {recall:.4f}, F1 = {f1:.4f}, AUC = {auc:.4f}")
```
